#### Shallow vs Deep Copy

-   **Shallow Copy**: Creates a new object, but populates it with *references* to the sub-objects found in the original. If you modify a mutable sub-object in the shallow copy, the original object will also be affected because they both refer to the same sub-object.
-   **Deep Copy**: Creates a new object and then, recursively, inserts *copies* into it of the sub-objects found in the original. Changes made to the deep copy do not affect the original object.

```python
arr = np.arange(1,11)
arr
```

```python
new_arr = arr #Shallow Copy
new_arr
```

```python
new_arr[0] = 99
new_arr
```

```python
arr #Affect Orignal Array
```

```python
new_arr = arr.copy() #Depth Copy
new_arr
```

```python
new_arr[9] = 70
new_arr
```

```python
arr #No Affect on Orignal
```

```python
import numpy as np

# Columns: [Age, Math Marks, Science Marks]
data = np.array([
    [18, 85, 78],   # Student 1
    [19, 92, 88],   # Student 2
    [17, 76, 95],   # Student 3
    [18, 65, 70],   # Student 4
    [20, 90, 85]    # Student 5
])
```

```python
#Get the shape of the matrix.

data.shape
```

```python
#Find the average age of students.

age = data[:6,0]
age.mean()
```

```python
#Extract Math marks of all students.

mathmarks= data[:6,1]
mathmarks
```

```python
#Find the highest Science mark.
score = data[:6,2]
score.max()
```

```python
#Get details of the student who scored more than 90 in Math.

std = mathmarks[mathmarks>90]
std
```

```python
#Increase Math marks of all students by 5.
mathmarks = mathmarks + 5
mathmarks
```

```python
#Find how many students are younger than 19.

age[age<19].size
```

```python
#Calculate the average marks in each subject (column-wise mean).

for i in range(1,3):
  print(data[:,i].mean())
```
